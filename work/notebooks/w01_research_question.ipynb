## 2. The question: decision, action, cost of a wrong call

*What decision does your work improve? Who acts on it? What does a wrong recommendation cost?*

### Research question

Which website pages should a content team review first because they show signs of decline or a useful opportunity for improvement?

- **Unit of analysis:** One webpage.
- **Output:** A ranked list of pages, with a priority score and readable reasons for each recommendation.
- **Decision improved:** Which pages should the content team spend its limited review time on first?
- **Who acts:** A content editor, SEO specialist, or content manager.
- **Possible action:** Review the page and decide whether to refresh, expand, protect, prune, or continue monitoring it.
- **Cost of a false positive:** The team may waste time reviewing or changing a page that did not need attention. An unnecessary change could also harm a page that was performing well.
- **Cost of a false negative:** The team may miss an important declining page and allow it to continue losing visibility, clicks, or traffic.
- **Why data or ML may help:** A team may have thousands of pages and many signals to consider. A data-supported ranking can combine these signals and help prioritize human review more consistently. The system would support the decision; it would not make the final editing decision automatically.


## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

```python
import pandas as pd

data_url = (
    "https://raw.githubusercontent.com/"
    "Abdelrahmanshaheen1/flyrank-ml-internship/"
    "main/data/raw/content_refresh_anonymized.csv"
)

df = pd.read_csv(data_url)

total_pages = len(df)

declining_mask = df["trend_direction"].str.lower().eq("down")
declining_pages = declining_mask.sum()
declining_rate = declining_mask.mean() * 100

visible_declining_mask = declining_mask & (df["impressions_90d"] >= 500)
visible_declining_pages = visible_declining_mask.sum()

print(f"Total pages in the starter dataset: {total_pages:,}")
print(
    f"Declining pages: {declining_pages:,} "
    f"({declining_rate:.1f}% of all pages)"
)
print(
    f"Declining pages with at least 500 impressions: "
    f"{visible_declining_pages:,}"
)
```

### Why these numbers support this lane

The starter dataset contains 30,000 webpages, which is too many for a content team to review manually. Of these pages, 16,262, or 54.2%, are labeled as declining. More importantly, 9,961 declining pages still received at least 500 impressions during the last 90 days. This suggests that many pages may still have meaningful visibility while losing performance. A ranked review queue could help the team focus first on declining pages where timely action may be most useful, instead of treating every page equally.
